# Reproducing Menon & Power (2024): all figures

[Menon & Power 2024](https://ui.adsabs.harvard.edu/abs/2024PASA...41...49M/abstract) (PASA, 41, e049; [arXiv:2405.03211](https://arxiv.org/abs/2405.03211)), "On bursty star formation during cosmological reionisation -- how does it influence the baryon mass content of dark matter halos?", is the paper **Ashvini**'s physics (star formation, delayed/instantaneous supernova feedback, UV-suppressed cosmological accretion, BH growth) implements. This notebook reproduces every figure in the paper (Figs. 1-8 and A1) using **pymctrees**-generated Monte Carlo merger trees (PCH08) run through Ashvini, with **10 000 trees per halo mass bin** (vs. the paper's 100) for smoother percentile statistics.

Fiducial parameters (paper Sec. 2, and already `run_params.yaml`'s defaults): $\epsilon_{\rm sf}=0.015$, $\epsilon_{\rm fb}=5$, $t^d=0.015$ Gyr, $z_{\rm rei}=7$. Cosmology matches the paper's stated Planck 2018 values via `pymctrees`' `config/menon_power_2024.yml` (see that file's comments for how its CAMB `As` was calibrated to $\sigma_8=0.815$).

**Known, disclosed differences from the paper** (see also each figure's own markdown below):

- Redshift sampling: the paper samples "equally spaced in the logarithm of the expansion factor" over $5\le z\le25$; this notebook uses pymctrees' native uniform-in-$z$ stepping ($dz=0.01$) over the same range.
- A halo's mass is held at 0 for any step before its tree becomes resolved (see `ashvini.pymctrees_adapter._mask_unresolved_prefix`) -- physically "this halo does not exist yet", not a growth artifact.
- Figures 3-8/A1 only ever plot median + 10th-90th percentile bands (matching the paper's own figures) -- per-halo trajectories for 10 000 halos aren't retained in memory; only Figs. 1 and 2 show individual/single-halo trajectories, as in the paper.
- Log-y axes can't display exactly-zero values (many halos are fully quenched under UV suppression, especially at low mass -- a real, checked model outcome, not a bug). Percentiles are clipped to a small fixed floor (1 $M_\odot$ for gas, 0.1 $M_\odot$ for stars) purely for display; a band sitting on the floor means "negligible/zero", not "exactly this floor value".
- Fig. 2 uses one representative halo (array index 0 of the 10 000) from the $10^7\,M_\odot$ sample, not necessarily the paper's own specific example halo -- same 4 scenarios, same physics.

Runtime: ~30-40 minutes (37 tree/forest generations at $N=10000$ for Figs. 1-8/A1, plus ~44 more for the black hole growth/feedback section below, an extension beyond the paper itself).

```python
import sys
import numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, '/Users/00075868/MyCodes/pymctrees/src')

from ashvini import configure, main as amain
from ashvini.pymctrees_adapter import build_forest_live

PYMCTREES_CONFIG = '/Users/00075868/MyCodes/pymctrees/config/menon_power_2024.yml'
N_HALOS = 10_000
Z0, Z_MAX, DZ = 5.0, 25.0, 0.01
SEED = 42
FIDUCIAL = dict(eps_sf=0.015, eps_fb=5, t_d=0.015, z_rei=7)

MPC_KM = 3.0856775814913673e19  # km per Mpc
GYR_S = 3.1556952e16  # s per Julian Gyr
T_RAW_TO_GYR = MPC_KM / GYR_S
```

```python
_tree_cache = {}
def get_forest(mass_bin):
    if mass_bin not in _tree_cache:
        _tree_cache[mass_bin] = build_forest_live(
            PYMCTREES_CONFIG, mass_bin, N_HALOS, Z0, Z_MAX, DZ,
            backend="numba", seed=SEED,
        )
    return _tree_cache[mass_bin]


def summarize(res, keep_fields=("gas_mass", "stars_mass"), example_idx=None):
    """
    Reduce a run_forest() result to what plotting needs: per-field 10th/50th
    /90th percentiles across halos at every timestep (Figs 3-8/A1 only ever
    plot percentile bands, matching the paper), plus cosmic_time/redshift,
    and optionally one example halo's full trajectory (Fig 2). Keeps
    memory at O(n_steps) rather than O(N_halos * n_steps) per scenario.
    """
    out = {"cosmic_time": res["cosmic_time"], "redshift": res["redshift"]}
    for field in keep_fields:
        with np.errstate(invalid="ignore"):
            out[field + "_p10"], out[field + "_p50"], out[field + "_p90"] = (
                np.nanpercentile(res[field], [10, 50, 90], axis=0)
            )
        out[field + "_nan_halos"] = int(np.isnan(res[field]).any(axis=1).sum())
    if example_idx is not None:
        out["example"] = {field: res[field][example_idx] for field in keep_fields}
    return out


def run_scenario(mass_bin, keep_fields=("gas_mass", "stars_mass"), example_idx=None, **overrides):
    params = dict(FIDUCIAL)
    params.update(overrides)
    configure.set_params(**params)
    halo_masses, halo_growth_rates, redshifts, _merger_mass = get_forest(mass_bin)
    result = amain.run_forest(halo_masses, halo_growth_rates, redshifts)
    return summarize(result, keep_fields=keep_fields, example_idx=example_idx)


def add_redshift_twin(ax, t_of_z_func):
    z_ticks = np.array([25, 15, 10, 8, 6, 5])
    t_ticks = t_of_z_func(z_ticks)
    ax2 = ax.twiny()
    ax2.set_xlim(ax.get_xlim())
    ax2.set_xticks(t_ticks)
    ax2.set_xticklabels([str(zt) for zt in z_ticks])
    ax2.set_xlabel('z')


FLOOR = {"gas_mass": 1.0, "stars_mass": 0.1}  # Msun; see markdown above re: log-axis floor clipping
_axis_positive_range = {}


def plot_percentile_pair(ax_gas, ax_stars, res, color, label, ls='-'):
    t = res['cosmic_time']
    for ax, field in [(ax_gas, 'gas_mass'), (ax_stars, 'stars_mass')]:
        floor = FLOOR[field]
        p10 = np.maximum(res[field + '_p10'], floor)
        p50 = np.maximum(res[field + '_p50'], floor)
        p90 = np.maximum(res[field + '_p90'], floor)
        ax.plot(t, p50, color=color, ls=ls, label=label if field == 'gas_mass' else None)
        ax.fill_between(t, p10, p90, color=color, alpha=0.15)
        ax.set_xlim(t.min(), t.max())

        finite_vals = np.concatenate([p10[np.isfinite(p10)], p50[np.isfinite(p50)], p90[np.isfinite(p90)]])
        if finite_vals.size:
            lo, hi = finite_vals.min(), finite_vals.max()
            prev = _axis_positive_range.get(id(ax))
            _axis_positive_range[id(ax)] = (lo, hi) if prev is None else (min(lo, prev[0]), max(hi, prev[1]))


def safe_log_yscale(ax):
    rng = _axis_positive_range.get(id(ax))
    if rng is not None:
        lo, hi = rng
        ax.set_yscale('log')
        ax.set_ylim(lo * 0.5, hi * 2)
```

## Figure 1: mass assembly of $10^8\,M_\odot$ halos

10 000 dark-matter-only trees, $M_0=10^8\,M_\odot$ at $z=5$ (100 shown individually for legibility). Pure `pymctrees` output, no Ashvini physics -- see `pymctrees/notebooks/pymctrees_demo.ipynb` §9 for the original single-figure version of this (with a detailed writeup of the resolution-floor/onset-redshift caveat) this notebook's Fig. 1 reuses the same approach for.

```python
import pymctrees.cosmo_utils as pymctrees_cosmo_utils
from pymctrees.utils import io as pymctrees_io

_pmt_params = pymctrees_io.get_params(PYMCTREES_CONFIG)
_pmt_cosmo = pymctrees_cosmo_utils.CosmoData(_pmt_params, redshift=[5.0])
t_of_z = lambda z: _pmt_cosmo.cosmic_time_at_z(z) * T_RAW_TO_GYR

halo_masses_1e8, _, z_1e8, _ = get_forest(1e8)
t_1e8 = t_of_z(z_1e8)
print(f"{N_HALOS} trees built; t = {t_1e8[0]:.3f}-{t_1e8[-1]:.3f} Gyr (z = {z_1e8[0]:.0f}-{z_1e8[-1]:.0f})")
```

```python
mass_masked = np.where(halo_masses_1e8 > 0, halo_masses_1e8, np.nan)
rate = np.gradient(halo_masses_1e8, t_1e8, axis=1) / 1e9  # Msun/yr
rate_masked = np.where(halo_masses_1e8 > 0, np.abs(rate), np.nan)

with np.errstate(invalid='ignore'):
    mass_p10, mass_p50, mass_p90 = np.nanpercentile(mass_masked, [10, 50, 90], axis=0)
    rate_p10, rate_p50, rate_p90 = np.nanpercentile(rate_masked, [10, 50, 90], axis=0)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
rng = np.random.default_rng(0)
plot_idx = rng.choice(N_HALOS, size=100, replace=False)

ax = axes[0]
for i in plot_idx:
    ax.plot(t_1e8, mass_masked[i], color='grey', lw=0.4, alpha=0.5)
ax.plot(t_1e8, mass_p50, color='red', lw=2, label='median')
ax.plot(t_1e8, mass_p10, color='red', lw=1, ls='--', label='10th/90th pct')
ax.plot(t_1e8, mass_p90, color='red', lw=1, ls='--')
ax.set_yscale('log')
ax.set_xlabel('cosmic time [Gyr]')
ax.set_ylabel(r'$M_{\rm halo}$ [M$_\odot$]')
ax.legend(fontsize=8)

ax = axes[1]
for i in plot_idx:
    ax.plot(t_1e8, rate_masked[i], color='grey', lw=0.4, alpha=0.5)
ax.plot(t_1e8, rate_p50, color='red', lw=2)
ax.plot(t_1e8, rate_p10, color='red', lw=1, ls='--')
ax.plot(t_1e8, rate_p90, color='red', lw=1, ls='--')
ax.set_yscale('log')
ax.set_xlabel('cosmic time [Gyr]')
ax.set_ylabel(r'$\dot{M}_{\rm halo}$ [M$_\odot$/yr]')

for ax in axes:
    add_redshift_twin(ax, t_of_z)
fig.suptitle(r'Fig 1: 10000 $M_0=10^8\,M_\odot$ ($z=5$) halos (100 shown)')
plt.tight_layout()
plt.show()
```

## Figure 8: stellar fraction vs. halo mass

11 halo mass bins, linearly spaced over $10^6\le m_h/M_\odot\le 10^{11}$ (per the paper's Sec. 2.2 -- note this spacing is *very* uneven on the log-mass axis the paper plots it on, since it's linear in $m_h$ itself: 10 of the 11 bins land between $10^{10}$ and $10^{11}\,M_\odot$). The $10^6\,M_\odot$ bin (leftmost) is **fully quenched** at all 3 epochs under this parameterisation (median and 10th percentile both exactly 0 -- see Fig. A1's flat left panel above) -- its stellar fraction is undefined ($0/0$) and dropped from the plot rather than shown as a spurious 0, which is why the x-axis below starts at $10^{10}\,M_\odot$, not $10^6\,M_\odot$.

```python
fig8_mass_bins = np.linspace(1e6, 1e11, 11)
res_fig8 = {}
for mass_bin in fig8_mass_bins:
    for sn_type in ["instantaneous", "delayed"]:
        key = (mass_bin, sn_type)
        if mass_bin == 1e6:
            res_fig8[key] = res_a1[(1e6, sn_type)]
        elif mass_bin == 1e11:
            res_fig8[key] = res_a1[(1e11, sn_type)]
        else:
            res_fig8[key] = run_scenario(mass_bin, sn_type=sn_type, uv_background=True)
print("done")
```

```python
epochs = [10, 7, 5]
colors = {"instantaneous": 'C0', "delayed": 'C3'}
markers = {10: 'o', 7: 's', 5: '^'}

fig, ax = plt.subplots(figsize=(7, 5.5))
for sn_type in ["instantaneous", "delayed"]:
    for z_epoch in epochs:
        fracs_p50 = []
        for mass_bin in fig8_mass_bins:
            res = res_fig8[(mass_bin, sn_type)]
            idx = np.argmin(np.abs(res['redshift'] - z_epoch))
            g50, s50 = res['gas_mass_p50'][idx], res['stars_mass_p50'][idx]
            denom50 = s50 + g50 if (s50 + g50) > 0 else np.nan
            fracs_p50.append(s50 / denom50)
        ax.plot(fig8_mass_bins, fracs_p50, color=colors[sn_type], marker=markers[z_epoch],
                ls='-' if sn_type == "delayed" else ':',
                label=f'{sn_type}, z={z_epoch}')
ax.set_xscale('log')
ax.set_xlabel(r'$m_{\rm halo}$ [M$_\odot$]')
ax.set_ylabel(r'$m_\star/(m_\star+m_{\rm gas})$')
ax.legend(fontsize=7, ncol=2)
fig.suptitle('Fig 8: stellar fraction vs halo mass')
plt.tight_layout()
plt.show()
```

## Black hole growth and feedback

Menon & Power (2024) doesn't model black holes at all -- this section is an extension, not a reproduction. Ashvini implements BH seeding (three channels), gas-supply-limited growth capped at a (possibly super-Eddington) multiple of the Eddington rate, a constant-$\eta_{\rm agn}$ AGN wind, and an optional isothermal-sphere M-sigma self-regulation coupling (King 2003, 2005; Power, Zubovas, Nayakshin & King 2011 -- PZNK11). See `MODELS.md`'s "Black hole growth"/"AGN feedback" sections for the full equations.

Four BH configurations are compared below:
- **no BH**: seeding disabled entirely (`bh_enabled=False`) -- the baseline every other figure in this notebook already implicitly used, since BH growth is *on* by default in `run_params.yaml` and `ashvini.configure` never touched it before this section.
- **const $\eta_{\rm agn}$**: the default constant-coupling AGN wind ($\eta_{\rm agn}=0.5$) -- this is what Figs. 1-8/A1 above actually ran with, whether or not that was obvious from the plots.
- **sigma wind only**: `sigma_feedback.enabled=True` -- replaces the constant coupling with one gated by $M_{\rm BH}/M_\sigma$, but (checked directly, see `tests/test_sigma_feedback.py`) does **not** cap $M_{\rm BH}$ by itself; it only changes the wind's strength.
- **sigma + hard cap**: adds `growth_cap_enabled=True` -- a hard cutoff at PZNK11's own predicted ceiling ($M_\sigma \times (1 + 0.41\,\sigma_{200}/h(z))$, their eq. 21-22). PZNK11 don't specify *how* growth halts (their words: "the most unclear part of any theory"), so the hard cutoff itself is a disclosed modelling choice grounded in their quantitative bound, not an equation read directly off the paper.

**A finding worth stating up front**: at Ashvini's fiducial BH efficiency ($\epsilon_{\rm BH}=0.001$), checked directly, none of these configurations differentiate at all -- BH growth stays purely Eddington-limited and never gets within four orders of magnitude of $M_\sigma$ for any halo in the mass range below, so the AGN feedback mechanisms have nothing to act on. This mirrors a real, known tension in the literature (pure-Eddington growth from small seeds is famously too slow to explain high-z SMBHs). The higher-mass comparison below uses a non-fiducial, literature-motivated ($\epsilon_{\rm BH}=0.05$, $f_{\rm Edd}=15$, "hyper-Eddington" early-growth scenarios -- e.g. Volonteri et al., Pacucci & Loeb) parameter choice so the comparison has something to actually show, the same spirit as Figs. 4-6's own non-fiducial sweeps.

### Does BH feedback matter at $10^7\,M_\odot$?

Simple on/off check at fiducial BH parameters, reusing the delayed+UV $10^7\,M_\odot$ tree already built for Figs. 2/3/4-7.

```python
from ashvini import black_holes_growth as bh_growth

res_1e7_bh = {}
res_1e7_bh["no_bh"] = run_scenario(1e7, sn_type="delayed", uv_background=True, bh_enabled=False, keep_fields=("gas_mass", "stars_mass", "bh_mass"))
res_1e7_bh["const_eta"] = run_scenario(1e7, sn_type="delayed", uv_background=True, bh_enabled=True, eta_agn=0.5, sigma_feedback_enabled=False, growth_cap_enabled=False, keep_fields=("gas_mass", "stars_mass", "bh_mass"))
print("done")
```

```python
FLOOR_BH = dict(FLOOR)
FLOOR_BH["bh_mass"] = 0.01

def plot_bh_triple(axes, res, color, label, ls='-'):
    t = res['cosmic_time']
    for ax, field in zip(axes, ['gas_mass', 'stars_mass', 'bh_mass']):
        floor = FLOOR_BH[field]
        p10 = np.maximum(res[field + '_p10'], floor)
        p50 = np.maximum(res[field + '_p50'], floor)
        p90 = np.maximum(res[field + '_p90'], floor)
        ax.plot(t, p50, color=color, ls=ls, label=label if field == 'gas_mass' else None)
        ax.fill_between(t, p10, p90, color=color, alpha=0.15)
        ax.set_yscale('log')
        ax.set_xlim(t.min(), t.max())

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
plot_bh_triple(axes, res_1e7_bh["no_bh"], 'grey', 'no BH', ls=':')
plot_bh_triple(axes, res_1e7_bh["const_eta"], 'C3', 'const eta_agn (BH on)', ls='-')
for ax, field in zip(axes, ['gas_mass', 'stars_mass', 'bh_mass']):
    ax.set_xlabel('cosmic time [Gyr]')
    ax.set_ylabel(f'{field} [Msun]')
axes[0].legend(fontsize=8)
fig.suptitle(r'$10^7\,M_\odot$ halos: BH on/off (delayed+UV, fiducial $\epsilon_{\rm BH}$)')
plt.tight_layout()
plt.show()
```

```python
configs = ['const_eta', 'sigma_wind_only', 'sigma_capped']
colors_bh = {'const_eta': 'C0', 'sigma_wind_only': 'C1', 'sigma_capped': 'C3'}
labels_bh = {'const_eta': 'const eta_agn (uncapped)', 'sigma_wind_only': 'sigma wind only (uncapped)', 'sigma_capped': 'sigma + hard cap'}

fig, ax = plt.subplots(figsize=(9, 6.5))
FLOOR_BH_MASS = 1e2
for bh_config in configs:
    p50s, p90s = [], []
    for mass_bin in fig8_mass_bins:
        res = res_fig8_bh[(mass_bin, bh_config)]
        idx = np.argmin(np.abs(res['redshift'] - 5))
        p50s.append(max(res['bh_mass_p50'][idx], FLOOR_BH_MASS))
        p90s.append(max(res['bh_mass_p90'][idx], FLOOR_BH_MASS))
    ax.plot(fig8_mass_bins, p50s, color=colors_bh[bh_config], lw=2, ls='-', label=f'{labels_bh[bh_config]}, median')
    ax.plot(fig8_mass_bins, p90s, color=colors_bh[bh_config], lw=1.5, ls=':', label=f'{labels_bh[bh_config]}, 90th pct')

ceilings = [max(bh_growth.growth_ceiling(m, 5.0), FLOOR_BH_MASS) for m in fig8_mass_bins]
ax.plot(fig8_mass_bins, ceilings, color='k', ls='--', lw=1.5, label=r'$M_\sigma$ ceiling (eq. 21-22)')

ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel(r'$m_{\rm halo}$ [M$_\odot$]')
ax.set_ylabel(r'$M_{\rm BH}$ [M$_\odot$] at z=5')
ax.legend(fontsize=8)
fig.suptitle(r'BH mass vs halo mass: uncapped vs capped self-regulation ($\epsilon_{\rm BH}=0.05$, $f_{\rm Edd}=15$)')
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
for bh_config in ['no_bh'] + configs:
    fracs = []
    for mass_bin in fig8_mass_bins:
        res = res_fig8_bh[(mass_bin, bh_config)]
        idx = np.argmin(np.abs(res['redshift'] - 5))
        g50, s50 = res['gas_mass_p50'][idx], res['stars_mass_p50'][idx]
        denom = s50 + g50 if (s50 + g50) > 0 else np.nan
        fracs.append(s50 / denom)
    c = 'grey' if bh_config == 'no_bh' else colors_bh[bh_config]
    lbl = 'no BH' if bh_config == 'no_bh' else labels_bh[bh_config]
    ax.plot(fig8_mass_bins, fracs, color=c, marker='o', label=lbl)
ax.set_xscale('log')
ax.set_xlabel(r'$m_{\rm halo}$ [M$_\odot$]')
ax.set_ylabel(r'$m_\star/(m_\star+m_{\rm gas})$ at z=5')
ax.legend(fontsize=8)
fig.suptitle('Stellar fraction vs halo mass: BH config comparison (delayed+UV)')
plt.tight_layout()
plt.show()
```

**Reading these two figures together**: the hard cap visibly holds the upper tail of $M_{\rm BH}$ at the $M_\sigma$ ceiling once halos are massive enough to seed and grow efficiently ($\gtrsim 2\times10^{10}\,M_\odot$ here), while the uncapped mechanisms (constant $\eta_{\rm agn}$ and the sigma wind-switch alone) let the 90th-percentile $M_{\rm BH}$ climb well past it -- direct confirmation that the hard cap does what the wind-only mechanisms don't (see `MODELS.md`'s caveat).

But this has **essentially no visible effect on the gas/stellar mass budget** -- the stellar-fraction curves for all four BH configurations (including no BH at all) are effectively on top of each other. Even a maximal AGN wind removes a negligible fraction of these halos' gas reservoirs compared to sustained cosmological accretion, the same "accretion dominates over feedback at high halo mass" pattern already seen for supernova feedback in Fig. 8 above. AGN feedback change the black hole population here without changing the galaxy population -- a real, if modest, result in its own right, not a null result to be explained away.

## Summary

All 9 figures from Menon & Power (2024) reproduced end-to-end using pymctrees-generated PCH08 trees (10 000 per bin) run through Ashvini's own implementation of the paper's physics. The qualitative trends match throughout: delayed feedback produces bursty/oscillatory gas-mass histories that instantaneous feedback cannot; higher $\epsilon_{\rm sf}$/weaker $\epsilon_{\rm fb}$/longer $t^d$/earlier $z_{\rm rei}$ each shift the baryon budget in the expected direction; low-mass halos are far more UV-suppressed than high-mass ones (Fig. A1); and stellar fraction rises with halo mass (Fig. 8, over the resolvable range). See the markdown notes above each figure, and the intro cell, for the specific disclosed differences from the paper's own methodology (sampling scheme, percentile-only statistics, and floor-clipping for display).

**Black hole growth and feedback** (an extension beyond the paper, not a reproduction): at fiducial BH efficiency, BH growth stays purely Eddington-limited and never approaches the $M_\sigma$ self-regulation scale for any halo here -- none of Ashvini's AGN feedback mechanisms differentiate. At elevated, literature-motivated ($\epsilon_{\rm BH}=0.05$, $f_{\rm Edd}=15$) BH growth, PZNK11's hard self-regulation cap visibly holds the upper tail of $M_{\rm BH}$ at its predicted ceiling where the wind-only mechanisms (constant $\eta_{\rm agn}$, or the sigma wind-switch alone) let it grow past -- but even this has no visible effect on gas/stellar mass: AGN feedback changes the black hole population without changing the galaxy population in this mass range, since cosmological accretion dominates over any of these wind strengths at halo masses where BH growth is significant.
